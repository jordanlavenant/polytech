```python
np.random.seed(42) # Reproductibilité

def randomVector(n):
    v = np.zeros(n)
    for i in range(n): v[i] = rd.randint(10)
    return v

C = randomVector(5)

print(C)
```

### 1.9 Parcourir une boucle `for` à l'envers
Construire le vecteur d'entiers `[n-1, ..., 1, 0]`.


```python
def reverseRandomVector(n):
    v = np.zeros(n)
    for i in range(n - 1, -1, -1): v[i] = n - i - 1
    return v

E = reverseRandomVector(5)

print(E)
```

### 1.11 Permuter deux lignes d'une matrice
Effectuer des permutations deux à deux et vérifier le résultat. On pourra utiliser la notation `B[[i1,i2],:]` pour désigner la matrice extraite de B composée des lignes `i1` et `i2`.

```python
def permute(M,  i1, i2):
    if (i1 > shape[0] - 1 or i2 > shape[1] - 1):
        return M
        
    M[[i1, i2], :] = M[[i2, i1], :]
    return M


np.random.seed(42)

B = randomMatrix(4)

print(B)

B = permute(B, 0, 3) # Première et dernière ligne

print(B)
```

## 2. Système triangulaire inférieur : descente (forward)
### 2.1 Rappeler la condition nécessaire et suffisante d'inversibilité
Pour une matrice triangulaire, expliciter la condition sur la diagonale.


Il ne faut pas que la matrice ait des $0$ sur la diagonale ($M_{ii} \neq 0$)

### 2.2 Coder `Descente(L, b)`
`L` est triangulaire inférieure inversible et `b` est le second membre.
La fonction retourne la solution `x` de `Lx=b`.


```python
def Descente(L, b):
    for i in range(np.shape(L)[0]):
        if L[i][i] == 0:
            raise ValueError("La matrice L n'est pas inversible")

    n = len(b)
    x = np.zeros(n)

    for i in range(n):
        somme = 0
        for j in range(i):
            somme += L[i][j] * x[j]

        x[i] = (b[i] - somme) / L[i][i]
    
    return x
```

## 3. Système triangulaire supérieur : montée (backward)
### 3.1 Coder `Montee(U, b)`
`U` est triangulaire supérieure inversible et `b` est le second membre.


```python
def Montee(U, b):
    for i in range(np.shape(U)[0]):
        if U[i][i] == 0:
            raise ValueError("La matrice U n'est pas inversible")

    n = len(b)
    x = np.zeros(n)

    for i in range(n - 1, -1, -1):
        somme = 0
        for j in range(i + 1, n):
            somme += U[i][j] * x[j]

        x[i] = (b[i] - somme) / U[i][i]
    
    return x
```

## 4. Factorisation LU
Dans chaque sous-partie, vérifier vos fonctions sur un exemple numérique en contrôlant l'erreur.


### 4.1 Version basique sans pivot (pas in place)
#### 4.1.1 Coder `MonLU(A)`
`A` est carrée. La fonction retourne `L` et `U` avec diagonale de `L` égale à 1.
Remarque : cette version peut échouer si un pivot est nul (ou très petit).


```python
def MonLU(A):
    n = A.shape[0]
    
    L = np.eye(n) # I
    U = np.zeros((n, n))
    
    for i in range(n):
        # Calcul de U (ligne i)
        for j in range(i, n):
            somme = 0
            for k in range(i):
                somme += L[i][k] * U[k][j]
            U[i][j] = A[i][j] - somme
        
        # Calcul de L (colonne i)
        for j in range(i + 1, n):
            somme = 0
            for k in range(i):
                somme += L[j][k] * U[k][i]
            
            if U[i][i] == 0:
                raise ValueError("Pivot nul, échec de LU sans pivot")
            
            L[j][i] = (A[j][i] - somme) / U[i][i]
    
    return L, U
```

Vérifier la décomposition sur un exemple de matrice carrée (`A` et `L@U`).


```python
np.random.seed(42)

A = randomMatrix(4) 
print(A, "\n")

L, U = MonLU(A)

print(L @ U)
```

#### 4.1.2 Écrire `SolveLU(A, b)`
La fonction retourne `x` en utilisant successivement descente puis montée.


```python
def SolveLU(A, b):
    L, U = MonLU(A)
    y = Descente(L, b)
    x = Montee(U, y)
    return x
```

Tester `SolveLU` sur un système aléatoire `Ax=b`.

```python
x = SolveLU(A, b)
print("x: \n", x)
```

```python
# devrait donner b
print("vérification: ", A @ x)
np.allclose(A @ x, b)
```

### 4.2 Version in place sans pivot


#### 4.2.1 Coder `MonLUInPlace(A)`
La fonction retourne `N` (in place) tel que `N = (L-I) + U` avec `A=LU` et diag(`L`)=1.


```python
def MonLUInPlace(A):
    n = A.shape[0]
    N = A.copy().astype(float)
    
    for i in range(n):
        # Calcul de U (ligne i)
        for j in range(i, n):
            somme = 0
            for k in range(i):
                somme += N[i][k] * N[k][j]
            N[i][j] = N[i][j] - somme
        
        # Vérification pivot
        if N[i][i] == 0:
            raise ValueError("Pivot nul, échec de LU sans pivot")
        
        # Calcul de L (colonne i)
        for j in range(i + 1, n):
            somme = 0
            for k in range(i):
                somme += N[j][k] * N[k][i]
            
            N[j][i] = (N[j][i] - somme) / N[i][i]
    
    return N
```

#### 4.2.2 Écrire `ExtractLU(N)`
Extraire `L` et `U` depuis `N`, avec ajout des 1 sur la diagonale de `L`.


```python
def ExtractLU(N):
    n = N.shape[0]
    L = np.eye(n) + np.tril(N, k=-1) # Diagonale = 1
    U = np.triu(N)
    return L, U
```

Vérifier la décomposition in place sur un exemple de matrice carrée.


```python
np.random.seed(42)

A = randomMatrix(SIZE)
print("A: \n", A, "\n")

N = MonLUInPlace(A)
print("N (L-U in place): \n", N, "\n")

L, U = ExtractLU(N)
print("L: \n", L, "\n")
print("U: \n", U, "\n")
```

#### 4.2.3 Résoudre un système linéaire à l'aide de `MonLUInPlace()`.


```python
print("L @ U: \n", L @ U, "\n")
print("A: \n", A, "\n")
```

### 4.3 Version in place avec pivot partiel (permutations)


#### 4.3.1 Coder `MonLUPivot(A)`
La fonction retourne `N` (in place) et `P`, matrice de permutation, telles que `PA = LU`.


```python
def MonLUPivot(A):
    n = A.shape[0]
    N = A.copy().astype(float)
    P = np.eye(n) # Permutation
    
    for i in range(n):
        # Pivot partiel
        pivot_index = np.argmax(np.abs(N[i:, i])) + i
        if N[pivot_index][i] == 0:
            raise ValueError("Matrice singulière, échec de LU avec pivot")
        
        # Permuter les lignes dans N et P
        permute(N, i, pivot_index)
        permute(P, i, pivot_index)

        # Calcul de U (ligne i)
        for j in range(i, n):
            somme = 0
            for k in range(i):
                somme += N[i][k] * N[k][j]
            N[i][j] = N[i][j] - somme
        
        # Calcul de L (colonne i)
        for j in range(i + 1, n):
            somme = 0
            for k in range(i):
                somme += N[j][k] * N[k][i]
            
            N[j][i] = (N[j][i] - somme) / N[i][i]
    
    return P, N
```

Vérifier la décomposition avec pivot sur un exemple de matrice carrée.


```python
np.random.seed(42)

A = randomMatrix(SIZE)

P, N = MonLUPivot(A)

PA = P @ A

L, U = ExtractLU(N)

LU = L @ U

np.allclose(PA, LU) # On vérifie PA = LU
```

#### 4.3.2 Résoudre un système linéaire à l'aide de `MonLUPivot()`.


```python
np.random.seed(42)

A = randomMatrix(SIZE)
print("A: \n", A, "\n")

P, N = MonLUPivot(A)

L, U = ExtractLU(N)
print("L: \n", L, "\n")
print("U: \n", U, "\n")

print("L @ U: \n", L @ U, "\n")
```
